# Adult Dataset Visualization (Set 2)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../DSBDALExam DataSets/Adult/adult.csv", header=None)

df.columns = [
    'age','workclass','fnlwgt','education','education-num',
    'marital-status','occupation','relationship','race','sex',
    'capital-gain','capital-loss','hours-per-week',
    'native-country','income'
]

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

df.head()
```

## Histogram

```python
print("Histogram Objective: To visualize distribution of Age")

plt.hist(df['age'], bins=20)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Pie Chart

```python
print("Pie Chart Objective: To show proportion of Income categories")

counts = df['income'].value_counts()
labels = counts.index

plt.pie(counts, labels=labels, autopct='%1.1f%%')
plt.title("Income Distribution")
plt.show()
```

## Box Plot

```python
print("Box Plot Objective: To detect outliers in Working Hours")

sns.boxplot(x=df['hours-per-week'])
plt.title("Box Plot of Hours per Week")
plt.xlabel("Hours per Week")
plt.show()
```

## Scatter Plot

```python
print("Scatter Plot Objective: To show relationship between Age and Working Hours")

plt.scatter(df['age'], df['hours-per-week'])
plt.title("Scatter Plot: Age vs Hours per Week")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.show()
```

## Combined Plot

```python
print("Combined Plot Objective: To show relationship and distribution together")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(df['age'], df['hours-per-week'])
axes[0].set_title("Scatter Plot")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Hours per Week")

sns.boxplot(y=df['hours-per-week'], ax=axes[1])
axes[1].set_title("Box Plot of Hours per Week")

plt.tight_layout()
plt.show()
```
