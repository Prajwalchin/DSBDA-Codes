# Iris Dataset ML
Data Cleaning, Outlier Removal, Transformation, Model Comparison

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

## Load Dataset (Fix Header)

```python
df = pd.read_csv("../DSBDALExam DataSets/Iris/Iris.csv", header=None)
df.columns = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Species']
df.head()
```

## Data Cleaning

```python
df.replace('?', np.nan, inplace=True)

for col in df.columns[:-1]:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.dropna(inplace=True)
df = df[(df.iloc[:, :-1] >= 0).all(axis=1)]
df.shape
```

## Outlier Removal

```python
Q1 = df.iloc[:, :-1].quantile(0.25)
Q3 = df.iloc[:, :-1].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df.iloc[:, :-1] < (Q1 - 1.5 * IQR)) | 
          (df.iloc[:, :-1] > (Q3 + 1.5 * IQR))).any(axis=1)]
df.shape
```

## Data Transformation

```python
df['Species'] = df['Species'].astype('category').cat.codes

X = df.drop('Species', axis=1)
y = df['Species']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## Model Comparison

```python
lr = LogisticRegression(max_iter=200)
lr.fit(X_train, y_train)
lr_acc = accuracy_score(y_test, lr.predict(X_test))

nb = GaussianNB()
nb.fit(X_train, y_train)
nb_acc = accuracy_score(y_test, nb.predict(X_test))

lr_acc, nb_acc
```
