# Adult Dataset Visualization
Visualization with objectives, labels and clean structure

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```python
df = pd.read_csv("../DSBDALExam DataSets/Adult/adult.csv", header=None)

df.columns = [
    'age','workclass','fnlwgt','education','education-num',
    'marital-status','occupation','relationship','race','sex',
    'capital-gain','capital-loss','hours-per-week',
    'native-country','income'
]

# Remove extra spaces
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

df.head()
```

## Histogram

```python
# Objective: To visualize distribution of Age
print("Histogram Objective: To visualize distribution of Age")

plt.hist(df['age'], bins=20)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Dot Plot

```python
# Objective: To display individual Age values
print("Dot Plot Objective: To display individual Age values")

plt.plot(df['age'], '.', color='blue')
plt.title("Dot Plot of Age")
plt.xlabel("Index")
plt.ylabel("Age")
plt.show()
```

## Bar Plot

```python
# Objective: To compare Male and Female count
print("Bar Plot Objective: To compare Male and Female count")

sns.countplot(x='sex', data=df)
plt.title("Bar Plot of Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

## Line Chart

```python
# Objective: To show average working hours for each age
print("Line Chart Objective: To show average Hours per Week for each Age")

df_avg = df.groupby('age')['hours-per-week'].mean()

plt.plot(df_avg.index, df_avg.values)
plt.title("Average Hours per Week vs Age")
plt.xlabel("Age")
plt.ylabel("Average Hours per Week")
plt.show()
```

## Box Plot + Histogram + Scatter Plot

```python
# Box Plot Objective: To detect outliers in working hours
print("Box Plot Objective: To detect outliers in Hours per Week")

sns.boxplot(x=df['hours-per-week'])
plt.title("Box Plot of Hours per Week")
plt.xlabel("Hours per Week")
plt.show()

# Histogram Objective: To visualize distribution of working hours
print("Histogram Objective: To show distribution of Hours per Week")

plt.hist(df['hours-per-week'], bins=20)
plt.title("Histogram of Hours per Week")
plt.xlabel("Hours per Week")
plt.ylabel("Frequency")
plt.show()

# Scatter Plot Objective: To show relationship between Age and Working Hours
print("Scatter Plot Objective: To show relationship between Age and Working Hours")

plt.scatter(df['age'], df['hours-per-week'])
plt.title("Scatter Plot: Age vs Hours per Week")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.show()
```
