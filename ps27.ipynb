# Steam Review Scraper
Fetch real-time reviews using Steam API

```python
import requests
import pandas as pd
```

## Fetch Data from Steam API

```python
# Steam API URL
url = "https://store.steampowered.com/appreviews/730?json=1"

# Send request
response = requests.get(url)

# Convert to JSON
data = response.json()

# Extract reviews
reviews = data['reviews']

print("Total Reviews Fetched:", len(reviews))
```

## Extract Review Information

```python
review_list = []

for review in reviews:

    username = review['author']['steamid']

    comment = review['review']

    rating = "Recommended" if review['voted_up'] else "Not Recommended"

    helpful = review['votes_up']

    funny = review['votes_funny']

    review_list.append({
        'Customer Name': username,
        'Review': comment,
        'Rating': rating,
        'Helpful Votes': helpful,
        'Funny Votes': funny
    })

print("Review data extracted successfully")
```

## Create DataFrame

```python
df = pd.DataFrame(review_list)

df.head()
```

## Save Data to CSV

```python
df.to_csv("steam_reviews.csv", index=False)

print("Steam reviews saved successfully!")
```
