# Heart Disease Visualization
All graphs with objectives and proper labeling

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../DSBDALExam DataSets/HeartDisease/Cleavland.csv", header=None)

df.columns = [
    'age','sex','cp','trestbps','chol','fbs','restecg',
    'thalach','exang','oldpeak','slope','ca','thal','num'
]

df.head()
```

## Histogram

```python
# Objective: To visualize distribution of Age
print("Histogram Objective: To visualize distribution of Age")

plt.hist(df['age'], bins=20)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Dot Plot

```python
# Objective: To display individual Age values
print("Dot Plot Objective: To display individual Age values")

plt.plot(df['age'], '.', color='blue')
plt.title("Dot Plot of Age")
plt.xlabel("Index")
plt.ylabel("Age")
plt.show()
```

## Bar Plot

```python
# Objective: To compare Chest Pain Type frequency
print("Bar Plot Objective: To compare Chest Pain Type frequency")

sns.countplot(x='cp', data=df)
plt.title("Bar Plot of Chest Pain Type")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")
plt.show()
```

## Line Chart

```python
# Fix: Sort by age
df_sorted = df.sort_values(by='age')

# Objective: Show trend between Age and Cholesterol
print("Line Chart Objective: To show trend between Age and Cholesterol")

plt.plot(df_sorted['age'], df_sorted['chol'])
plt.title("Line Chart: Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()
```

## Box Plot + Histogram + Scatter Plot

```python
# Box Plot Objective: To detect outliers in Cholesterol
print("Box Plot Objective: To detect outliers in Cholesterol")

sns.boxplot(x=df['chol'])
plt.title("Box Plot of Cholesterol")
plt.xlabel("Cholesterol")
plt.show()

# Histogram Objective: To show distribution of Cholesterol
print("Histogram Objective: To show distribution of Cholesterol")

plt.hist(df['chol'], bins=20)
plt.title("Histogram of Cholesterol")
plt.xlabel("Cholesterol")
plt.ylabel("Frequency")
plt.show()

# Scatter Plot Objective: To show relationship between Age and Cholesterol
print("Scatter Plot Objective: To show relationship between Age and Cholesterol")

plt.scatter(df['age'], df['chol'])
plt.title("Scatter Plot: Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()
```
