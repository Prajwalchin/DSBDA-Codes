# Heart Disease Visualization
All graphs with objectives and proper labeling

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../DSBDALExam DataSets/HeartDisease/Cleavland.csv", header=None)

df.columns = [
    'age','sex','cp','trestbps','chol','fbs','restecg',
    'thalach','exang','oldpeak','slope','ca','thal','num'
]

df.head()
```

## Histogram

```python
# Objective: To visualize distribution of Age
print("Histogram Objective: To visualize distribution of Age")

plt.hist(df['age'],bins=20)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Dot Plot

```python
# Objective: To display individual Age values
print("Dot Plot Objective: To display individual Age values")

plt.plot(df['age'], '.', color='blue')
plt.title("Dot Plot of Age")
plt.xlabel("Index")
plt.ylabel("Age")
plt.show()
```

## Bar Plot

```python
# Objective: To compare Chest Pain Type frequency
print("Bar Plot Objective: To compare Chest Pain Type frequency")

sns.countplot(x='cp', data=df)
plt.title("Bar Plot of Chest Pain Type")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")
plt.show()
```

## Line Chart

```python
# Compute average cholesterol for each age
avg_chol_by_age = (
    df.groupby('age', as_index=False)['chol'].mean()
)

# Objective: Show average cholesterol for each age
print("Line Chart Objective: To show average Cholesterol for each Age")

plt.plot(avg_chol_by_age['age'], avg_chol_by_age['chol'], marker='o', markersize=4)
plt.title("Line Chart: Average Cholesterol by Age")
plt.xlabel("Age")
plt.ylabel("Average Cholesterol")
plt.show()
```

## Box Plot + Histogram + Scatter Plot

```python
plt.figure(figsize=(15, 4))

# Histogram
plt.subplot(1, 3, 1)
plt.hist(df['age'], bins=20)
plt.title("Histogram")

# Scatter Plot
plt.subplot(1, 3, 2)
plt.scatter(df['age'], df['chol'])
plt.title("Scatter Plot")

# Box Plot
plt.subplot(1, 3, 3)
sns.boxplot(y=df['chol'])
plt.title("Box Plot")

plt.tight_layout()
plt.show()
```
